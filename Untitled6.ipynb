```python
# Simple Medical Expert System
def expert_system():
    print("--- Welcome to AI Health Advisor ---")
    print("Please answer with 'yes' or 'no'\n")

    # Questions (Symptoms)
    fever = input("Do you have a fever? ").lower()
    cough = input("Do you have a cough? ").lower()
    headache = input("Do you have a headache? ").lower()

    # Rules (Logic)
    if fever == "yes" and cough == "yes":
        result = "Possibility of Common Cold or Flu. Please rest and drink fluids."
    elif fever == "yes" and headache == "yes":
        result = "Possibility of Heatstroke or Infection. Please consult a doctor."
    elif cough == "yes":
        result = "It might be a throat irritation. Try some honey and tea."
    else:
        result = "Symptoms are unclear. Stay healthy and keep monitoring!"

    print("\n--- Diagnosis Result ---")
    print(result)

# Run the system
expert_system()
```
